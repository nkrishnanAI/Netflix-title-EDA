# Netflix Show Data Analysis

## Project objective
The objective of this project is to find insights for content growth,regional distribution and target audience, and how netflix content has been evolved through the year.

### Importing important libraries to perform analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading Dataset

```python
df = pd.read_csv("netflix_titles.csv")
df.head(4)
```

Making minor corrections for better readability

```python
df.rename(columns={'show_id' :'Show_Number','title':'Title','type':'Show_Type','director':'Director','cast':'Cast','country':'Country','date_added':'Date_added','release_year':'Release_Year','rating':'Rating','duration':'Duration','listed_in':'Listed_In','description':'Description'}, inplace=True)
df.head(4)
```

## Data cleaning and handling the missing values.

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df['Country'] = df['Country'].fillna('Unknown')
df['Director'] = df['Director'].fillna('Unknown')
df['Cast'] = df['Cast'].fillna('Unknown')
df = df.dropna(subset=['Rating'])
```

```python
df.isnull().sum()
```

```python
df['Date_added'] = pd.to_datetime(df['Date_added'])
df.info()
```

```python
df['Rating'] = df.groupby('Show_Type')['Rating'].transform(lambda x: x.fillna(x.mode()[0]))
df.isnull().sum()
```

```python
df = df.dropna(subset=['Duration'])
df = df.dropna(subset=['Date_added'])
df = df.dropna(subset=['Rating'])
df['Duration_int'] = df['Duration'].str.extract('(\\d+)').astype(int)
df.head()
```

```python
df.isnull().sum()
```

```python
df['Year_Added'] = df['Date_added'].dt.year

content_per_year = df['Year_Added'].value_counts().sort_index()

plt.figure(figsize=(10,5))
content_per_year.plot()
plt.title("Netflix Content Added Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
sns.countplot(data=df, x='Show_Type')
plt.title("Movies vs TV Shows on Netflix")
plt.show()
```

```python
pd.crosstab(df['Year_Added'], df['Show_Type']).plot(kind='bar', stacked=True, figsize=(12,6))
plt.title("Movies vs TV Shows Added Over Time")
plt.show()
```

```python
top_countries = df['Country'].value_counts().head(10)

sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title("Top 10 Content Producing Countries")
plt.xlabel("Number of Titles")
plt.show()
```

```python
genres = df['Listed_In'].str.split(', ').explode()
top_genres = genres.value_counts().head(10)

top_genres.plot(kind='bar', figsize=(10,5))
plt.title("Top 10 Genres on Netflix")
plt.ylabel("Number of Titles")
plt.show()
```

```python
sns.histplot(df[df['Show_Type']=='Movie']['Duration_int'], bins=30)
plt.title("Movie Duration Distribution")
plt.xlabel("Duration (minutes)")
plt.show()
```
