# Netflix Show Data Analysis

## Project objective
#### The objective of this project is to find insights for content growth,regional distribution and target audience, and how netflix content has been evolved through the year.

### Importing important libraries to perform analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading Dataset

```python
df = pd.read_csv("netflix_titles.csv")
df.head(4)
```

Making minor corrections for better readability

```python
df.rename(columns={'show_id' :'Show_Number','title':'Title','type':'Show_Type','director':'Director','cast':'Cast','country':'Country','date_added':'Date_added','release_year':'Release_Year','rating':'Rating','duration':'Duration','listed_in':'Listed_In','description':'Description'}, inplace=True)
df.head(4)
```

## Data cleaning and handling the missing values.

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df['Country'] = df['Country'].fillna('Unknown')
df['Director'] = df['Director'].fillna('Unknown')
df['Cast'] = df['Cast'].fillna('Unknown')
df = df.dropna(subset=['Rating'])
```

```python
df.isnull().sum()
```

```python
df['Date_added'] = pd.to_datetime(df['Date_added'])
df.info()
```

```python
df['Rating'] = df.groupby('Show_Type')['Rating'].transform(lambda x: x.fillna(x.mode()[0]))
df.isnull().sum()
```

```python
df = df.dropna(subset=['Duration'])
df = df.dropna(subset=['Date_added'])
df = df.dropna(subset=['Rating'])
df['Duration_int'] = df['Duration'].str.extract('(\\d+)').astype(int)
df.head()
```
